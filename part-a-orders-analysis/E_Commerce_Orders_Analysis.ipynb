# E-Commerce Orders & Reviews

# Part-A: Objectives
1) Clean inconsistent product and category names
2) Handle missing category values
3) Calculate total revenue per category
4) Identify the top 2 best-selling products by quantity

```python
import pandas as pd

# Original dataset
orders=[
    {"order_id": "O1", "product": "Wireless Mouse",
     "category": "electronics", "qty": 3, "unit_price": 20.0},

    {"order_id": "O2", "product": "wireless mouse",
     "category": "Electronics", "qty": 2, "unit_price": 20.0},

    {"order_id": "O3", "product": "Yoga Mat",
     "category": "", "qty": 5, "unit_price": 15.0},

    {"order_id": "O4", "product": "Desk Lamp",
     "category": "Home", "qty": 1, "unit_price": 25.0},

    {"order_id": "O5", "product": "Yoga Mat",
     "category": "Fitness", "qty": 2, "unit_price": 15.0}
]

df=pd.DataFrame(orders)
df["product"]=df["product"].str.strip().str.title()
df["category"]=df["category"].str.strip().str.title()

df.loc[
    (df["product"]=="Yoga Mat") & (df["category"]==""),
    "category"
]="Fitness"

df["revenue"]=df["qty"]*df["unit_price"]

print("Cleaned Dataset")
print(df.to_string(index=False))

category_revenue=df.groupby("category")["revenue"].sum()

print("\nRevenue Per Category")
print(category_revenue)

# Top 2 products by quantity
top_products=(
    df.groupby("product")["qty"].sum().sort_values(ascending=False).head(2)
)

print("\nTop 2 Best-Selling Products")
print(top_products)
```
